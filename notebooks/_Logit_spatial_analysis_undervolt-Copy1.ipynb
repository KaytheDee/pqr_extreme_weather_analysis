### Mean collapsing; hide y-axis; make points more central (ref should be same) 

```python
def plot_site_or_distances_from_reference_binned(
    gdf,
    base_gdf=None,
    use_base_df=False,
    ref_easting=None,
    ref_northing=None,
    use_union_centroid=True,
    figsize=(9, 7),
    x_label_fontsize=15,
    x_labelpad=15,
    x_tick_fontsize=12,
    y_label=None,
    y_label_fontsize=15,
    y_labelpad=15,
    y_tick_fontsize=12,
    y_rotation=0,
    show_y_axis=True,

    # Sequential colors
    colors=None,

    # Binning
    n_bins=4,
    bins=None,
    bin_round=2,
    use_bins=False,

    # Plot styling
    point_size=70,
    point_alpha=0.9,
    point_edgecolor='none',
    point_linewidth=0.0,
    annotate=False,
    cbar_label='Undervoltage Likelihood',
    cbar_labelpad=20,
    cbar_label_fontsize=15,
    cbar_tick_fontsize=10.5,
    x_rotation=0,

    # Background basemap
    show_basemap=False,
    basemap_facecolor='none',
    basemap_edgecolor='lightgrey',
    basemap_linewidth=0.4,
    basemap_alpha=1.0,

    # Reference lines
    show_vertical_reference_line=True,
    show_horizontal_reference_line=True,
    reference_linecolor='black',
    reference_linestyle='--',
    reference_linewidth=1,
    reference_alpha=0.6,

    # Panel label using ax.transAxes
    panel_label=None,
    panel_label_x=0.02,
    panel_label_y=0.98,
    panel_label_fontsize=16,
    panel_label_fontweight='bold',
    panel_label_ha='left',
    panel_label_va='top',

    # Distance options
    use_km=True,
    x_ref_offset=0,
    y_ref_offset=0,

    # Axis limits
    xlim=None,
    ylim=None,

    # Site collapsing
    collapse_sites=True,
    collapse_method='max',

    # Diverging OR cmap
    use_diverging_or_cmap=True,
    or_reference=1,

    # Colormap options
    cmap_name='RdBu_r',
    use_custom_colors=False,
    diverging_colors=None,

    # Normalization controls
    symmetric_cmap=True,
    vmin=None,
    vmax=None,

    # Colorbar ticks
    cbar_ticks=None,

    # Layout
    colorbar_fraction=0.046,
    colorbar_pad=0.04,

    # Diagnostics
    diagnostics=True,

    # Saving
    save_path=None,
    dpi=300
):
    """
    Plot site-level odds ratios using continuous or binned colors.

    Locations are expressed relative to a selected reference point.
    A polygon basemap can optionally be transformed into the same
    relative-coordinate system.

    Reference-point priority
    ------------------------
    1. If use_base_df=True, use the union centroid of base_gdf.
    2. Otherwise, if use_union_centroid=True, use the union centroid
       of the plotted site geometries.
    3. Otherwise, use ref_easting and ref_northing when supplied.
       Any missing coordinate is replaced by the mean site coordinate.

    Returns
    -------
    GeoDataFrame
        Processed site data containing easting, northing, x_distance,
        and y_distance.
    """

    import numpy as np
    import geopandas as gpd
    import matplotlib.pyplot as plt

    from shapely import affinity

    from matplotlib.colors import (
        BoundaryNorm,
        LinearSegmentedColormap,
        ListedColormap,
        Normalize,
        TwoSlopeNorm
    )

    # -------------------------------------------------
    # Default colors
    # -------------------------------------------------
    if colors is None:
        colors = [
            '#fee5d9',
            '#fcae91',
            '#fb6a4a',
            '#cb181d'
        ]

    if diverging_colors is None:
        diverging_colors = [
            '#2166ac',
            '#67a9cf',
            '#f7f7f7',
            '#f4a582',
            '#b2182b'
        ]

    # -------------------------------------------------
    # Prepare site GeoDataFrame
    # -------------------------------------------------
    input_crs = getattr(
        gdf,
        'crs',
        None
    )

    gdf = gpd.GeoDataFrame(
        gdf.copy(),
        geometry='geometry',
        crs=input_crs
    )

    if gdf.crs is None:
        gdf = gdf.set_crs(
            epsg=32630
        )

    if gdf.crs.is_geographic:
        raise ValueError(
            'gdf must use a projected CRS before calculating '
            'distances. For Greater Accra, use '
            'gdf.to_crs(epsg=32630).'
        )

    required_columns = {
        'site_id',
        'or_exposure',
        'geometry'
    }

    missing_columns = (
        required_columns
        - set(gdf.columns)
    )

    if missing_columns:
        raise ValueError(
            f'Missing columns in gdf: '
            f'{sorted(missing_columns)}'
        )

    # -------------------------------------------------
    # Clean invalid rows
    # -------------------------------------------------
    gdf['or_exposure'] = (
        gdf['or_exposure']
        .astype(float)
    )

    gdf = gdf.dropna(
        subset=[
            'site_id',
            'or_exposure',
            'geometry'
        ]
    ).copy()

    gdf = gdf[
        ~gdf.geometry.is_empty
    ].copy()

    gdf = gdf[
        gdf.geometry.is_valid
    ].copy()

    gdf = gdf[
        np.isfinite(
            gdf['or_exposure']
        )
    ].copy()

    gdf = gdf[
        gdf['or_exposure'] > 0
    ].copy()

    if gdf.empty:
        raise ValueError(
            'No valid rows remain after removing missing, '
            'infinite, non-positive, or invalid observations.'
        )

    # -------------------------------------------------
    # Collapse multiple rows per site
    # -------------------------------------------------
    if collapse_sites:
        if collapse_method == 'max':
            idx = (
                gdf.groupby(
                    'site_id'
                )['or_exposure']
                .idxmax()
            )

            gdf = (
                gdf.loc[idx]
                .copy()
                .reset_index(drop=True)
            )

        elif collapse_method == 'min':
            idx = (
                gdf.groupby(
                    'site_id'
                )['or_exposure']
                .idxmin()
            )

            gdf = (
                gdf.loc[idx]
                .copy()
                .reset_index(drop=True)
            )

        elif collapse_method == 'mean':
            original_crs = gdf.crs

            agg_dict = {
                'or_exposure': (
                    'or_exposure',
                    'mean'
                ),
                'geometry': (
                    'geometry',
                    'first'
                )
            }

            optional_columns = [
                'cesi_level',
                'site_latitude',
                'site_longitude'
            ]

            for column in optional_columns:
                if column in gdf.columns:
                    agg_dict[column] = (
                        column,
                        'first'
                    )

            gdf = (
                gdf.groupby(
                    'site_id',
                    as_index=False
                )
                .agg(**agg_dict)
            )

            gdf = gpd.GeoDataFrame(
                gdf,
                geometry='geometry',
                crs=original_crs
            )

        else:
            raise ValueError(
                "collapse_method must be 'max', "
                "'min', or 'mean'."
            )

    # -------------------------------------------------
    # Site coordinates
    # -------------------------------------------------
    centroid_gdf = gdf.copy()

    centroid_gdf['easting'] = (
        centroid_gdf.geometry.x
    )

    centroid_gdf['northing'] = (
        centroid_gdf.geometry.y
    )

    # -------------------------------------------------
    # Prepare basemap when needed
    # -------------------------------------------------
    prepared_base_gdf = None

    if use_base_df or show_basemap:
        if base_gdf is None:
            raise ValueError(
                'base_gdf must be provided when '
                'use_base_df=True or show_basemap=True.'
            )

        base_crs = getattr(
            base_gdf,
            'crs',
            None
        )

        prepared_base_gdf = gpd.GeoDataFrame(
            base_gdf.copy(),
            geometry='geometry',
            crs=base_crs
        )

        if prepared_base_gdf.crs is None:
            prepared_base_gdf = (
                prepared_base_gdf.set_crs(
                    gdf.crs
                )
            )

        if prepared_base_gdf.crs != gdf.crs:
            prepared_base_gdf = (
                prepared_base_gdf.to_crs(
                    gdf.crs
                )
            )

        prepared_base_gdf = (
            prepared_base_gdf
            .dropna(subset=['geometry'])
            .copy()
        )

        prepared_base_gdf = prepared_base_gdf[
            ~prepared_base_gdf.geometry.is_empty
        ].copy()

        prepared_base_gdf = prepared_base_gdf[
            prepared_base_gdf.geometry.is_valid
        ].copy()

        if prepared_base_gdf.empty:
            raise ValueError(
                'No valid geometries were found '
                'in base_gdf.'
            )

    # -------------------------------------------------
    # Determine reference point
    # -------------------------------------------------
    if use_base_df:
        try:
            base_geom = (
                prepared_base_gdf
                .geometry
                .union_all()
            )

        except Exception:
            base_geom = (
                prepared_base_gdf
                .geometry
                .unary_union
            )

        reference_point = (
            base_geom.centroid
        )

        ref_easting = (
            reference_point.x
        )

        ref_northing = (
            reference_point.y
        )

        reference_source = (
            'base_gdf union centroid'
        )

    elif use_union_centroid:
        try:
            combined_geom = (
                centroid_gdf
                .geometry
                .union_all()
            )

        except Exception:
            combined_geom = (
                centroid_gdf
                .geometry
                .unary_union
            )

        reference_point = (
            combined_geom.centroid
        )

        ref_easting = (
            reference_point.x
        )

        ref_northing = (
            reference_point.y
        )

        reference_source = (
            'site union centroid'
        )

    else:
        if ref_easting is None:
            ref_easting = (
                centroid_gdf['easting']
                .mean()
            )

        if ref_northing is None:
            ref_northing = (
                centroid_gdf['northing']
                .mean()
            )

        reference_source = (
            'supplied coordinates or site means'
        )

    # -------------------------------------------------
    # Apply reference offsets
    # -------------------------------------------------
    if use_km:
        conversion_factor = 1000
        unit_label = 'km'

    else:
        conversion_factor = 1
        unit_label = 'm'

    ref_easting += (
        x_ref_offset
        * conversion_factor
    )

    ref_northing += (
        y_ref_offset
        * conversion_factor
    )

    # -------------------------------------------------
    # Calculate relative distances
    # -------------------------------------------------
    centroid_gdf['x_distance'] = (
        centroid_gdf['easting']
        - ref_easting
    ) / conversion_factor

    centroid_gdf['y_distance'] = (
        centroid_gdf['northing']
        - ref_northing
    ) / conversion_factor

    x_label = (
        f'East–West distance from reference '
        f'({unit_label})'
    )

    # -------------------------------------------------
    # Data and plotting ranges
    # -------------------------------------------------
    or_min = (
        centroid_gdf['or_exposure']
        .min()
    )

    or_max = (
        centroid_gdf['or_exposure']
        .max()
    )

    if (
        symmetric_cmap
        and vmin is None
        and vmax is None
    ):
        distance_below = abs(
            or_reference - or_min
        )

        distance_above = abs(
            or_max - or_reference
        )

        max_distance = max(
            distance_below,
            distance_above
        )

        vmin_plot = max(
            or_reference - max_distance,
            0
        )

        vmax_plot = (
            or_reference + max_distance
        )

    else:
        vmin_plot = (
            vmin
            if vmin is not None
            else or_min
        )

        vmax_plot = (
            vmax
            if vmax is not None
            else or_max
        )

    if vmin_plot >= vmax_plot:
        raise ValueError(
            f'Invalid color scale: '
            f'vmin_plot={vmin_plot}, '
            f'vmax_plot={vmax_plot}.'
        )

    # -------------------------------------------------
    # Colormap and normalization
    # -------------------------------------------------
    if (
        use_diverging_or_cmap
        and not use_bins
    ):
        if use_custom_colors:
            cmap = (
                LinearSegmentedColormap.from_list(
                    'custom_or_diverging_cmap',
                    diverging_colors
                )
            )

        else:
            cmap = plt.get_cmap(
                cmap_name
            )

        if (
            vmin_plot
            < or_reference
            < vmax_plot
        ):
            norm = TwoSlopeNorm(
                vmin=vmin_plot,
                vcenter=or_reference,
                vmax=vmax_plot
            )

        else:
            norm = Normalize(
                vmin=vmin_plot,
                vmax=vmax_plot
            )

            if diagnostics:
                print(
                    f'Warning: OR reference '
                    f'{or_reference} is outside the '
                    f'plotting range '
                    f'({vmin_plot:.3f}, '
                    f'{vmax_plot:.3f}).'
                )

    elif use_bins:
        if bins is None:
            bins_plot = np.linspace(
                vmin_plot,
                vmax_plot,
                n_bins + 1
            )

            bins_plot = np.round(
                bins_plot,
                bin_round
            )

            bins_plot = np.unique(
                bins_plot
            )

        else:
            bins_plot = np.asarray(
                bins,
                dtype=float
            )

        if len(bins_plot) < 2:
            raise ValueError(
                'At least two distinct bin boundaries '
                'are required.'
            )

        number_of_intervals = (
            len(bins_plot) - 1
        )

        if len(colors) < number_of_intervals:
            raise ValueError(
                f'{number_of_intervals} bin colors are '
                f'required, but only {len(colors)} '
                f'were supplied.'
            )

        cmap = ListedColormap(
            colors[:number_of_intervals]
        )

        norm = BoundaryNorm(
            bins_plot,
            cmap.N,
            clip=True
        )

    else:
        cmap = (
            LinearSegmentedColormap.from_list(
                'smooth_custom_cmap',
                colors
            )
        )

        norm = Normalize(
            vmin=vmin_plot,
            vmax=vmax_plot
        )

    # -------------------------------------------------
    # Diagnostics
    # -------------------------------------------------
    if diagnostics:
        print(
            '\n--- OR plot diagnostics ---'
        )

        print(
            'Rows after cleaning/collapse:',
            len(centroid_gdf)
        )

        print(
            'Unique sites:',
            centroid_gdf[
                'site_id'
            ].nunique()
        )

        print(
            'CRS:',
            centroid_gdf.crs
        )

        print(
            'Reference source:',
            reference_source
        )

        print(
            'Reference easting:',
            ref_easting
        )

        print(
            'Reference northing:',
            ref_northing
        )

        print(
            'Reference offset:',
            f'x={x_ref_offset} {unit_label}, '
            f'y={y_ref_offset} {unit_label}'
        )

        print(
            'x range:',
            centroid_gdf[
                'x_distance'
            ].min(),
            centroid_gdf[
                'x_distance'
            ].max()
        )

        print(
            'y range:',
            centroid_gdf[
                'y_distance'
            ].min(),
            centroid_gdf[
                'y_distance'
            ].max()
        )

        print(
            'OR range:',
            or_min,
            or_max
        )

        print(
            'Color range:',
            vmin_plot,
            vmax_plot
        )

        print(
            'xlim:',
            xlim
        )

        print(
            'ylim:',
            ylim
        )

        print(
            'Show basemap:',
            show_basemap
        )

        print(
            'Show vertical reference line:',
            show_vertical_reference_line
        )

        print(
            'Show horizontal reference line:',
            show_horizontal_reference_line
        )

        print(
            '---------------------------\n'
        )

    # -------------------------------------------------
    # Create figure and axes
    # -------------------------------------------------
    fig, ax = plt.subplots(
        figsize=figsize
    )

    # -------------------------------------------------
    # Optional basemap in relative coordinates
    # -------------------------------------------------
    if show_basemap:
        relative_base = (
            prepared_base_gdf.copy()
        )

        relative_base['geometry'] = (
            relative_base.geometry.apply(
                lambda geometry: affinity.scale(
                    affinity.translate(
                        geometry,
                        xoff=-ref_easting,
                        yoff=-ref_northing
                    ),
                    xfact=(
                        1 / conversion_factor
                    ),
                    yfact=(
                        1 / conversion_factor
                    ),
                    origin=(0, 0)
                )
            )
        )

        relative_base.plot(
            ax=ax,
            facecolor=basemap_facecolor,
            edgecolor=basemap_edgecolor,
            linewidth=basemap_linewidth,
            alpha=basemap_alpha,
            zorder=1
        )

    # -------------------------------------------------
    # Scatter plot
    # -------------------------------------------------
    sc = ax.scatter(
        centroid_gdf['x_distance'],
        centroid_gdf['y_distance'],
        c=centroid_gdf['or_exposure'],
        cmap=cmap,
        norm=norm,
        s=point_size,
        alpha=point_alpha,
        edgecolors=point_edgecolor,
        linewidths=point_linewidth,
        zorder=3
    )

    # -------------------------------------------------
    # Reference lines
    # -------------------------------------------------
    if show_vertical_reference_line:
        ax.axvline(
            0,
            color=reference_linecolor,
            linestyle=reference_linestyle,
            linewidth=reference_linewidth,
            alpha=reference_alpha,
            zorder=2
        )

    if show_horizontal_reference_line:
        ax.axhline(
            0,
            color=reference_linecolor,
            linestyle=reference_linestyle,
            linewidth=reference_linewidth,
            alpha=reference_alpha,
            zorder=2
        )

    # -------------------------------------------------
    # Panel label
    # -------------------------------------------------
    if panel_label is not None:
        ax.text(
            panel_label_x,
            panel_label_y,
            str(panel_label),
            transform=ax.transAxes,
            fontsize=panel_label_fontsize,
            fontweight=panel_label_fontweight,
            horizontalalignment=panel_label_ha,
            verticalalignment=panel_label_va,
            clip_on=False,
            zorder=10
        )

    # -------------------------------------------------
    # Site annotations
    # -------------------------------------------------
    if annotate:
        for _, row in centroid_gdf.iterrows():
            ax.text(
                row['x_distance'],
                row['y_distance'],
                str(row['site_id']),
                fontsize=8,
                horizontalalignment='left',
                verticalalignment='bottom',
                zorder=4
            )

    # -------------------------------------------------
    # Colorbar
    # -------------------------------------------------
    cbar = fig.colorbar(
        sc,
        ax=ax,
        fraction=colorbar_fraction,
        pad=colorbar_pad
    )

    cbar.set_label(
        cbar_label,
        fontsize=cbar_label_fontsize,
        fontweight='bold',
        labelpad=cbar_labelpad
    )

    cbar.ax.tick_params(
        labelsize=cbar_tick_fontsize
    )

    if cbar_ticks is not None:
        cbar.set_ticks(
            cbar_ticks
        )

    cbar.ax.yaxis.set_major_formatter(
        plt.FuncFormatter(
            lambda value, _:
            f'{value:g}X'
        )
    )

    # -------------------------------------------------
    # Axes
    # -------------------------------------------------
    ax.set_xlabel(
        x_label,
        fontsize=x_label_fontsize,
        fontweight='bold',
        labelpad=x_labelpad
    )

    if y_label is None:
        y_label = (
            'North-South Distance from Reference (km)'
            if use_km
            else 'North-South Distance from Reference (m)'
        )

    ax.yaxis.set_visible(
        show_y_axis
    )

    if show_y_axis:
        ax.set_ylabel(
            y_label,
            fontsize=y_label_fontsize,
            fontweight='bold',
            labelpad=y_labelpad
        )

    ax.spines['top'].set_visible(
        False
    )

    ax.spines['right'].set_visible(
        False
    )

    ax.spines['left'].set_visible(
        show_y_axis
    )

    ax.spines['bottom'].set_visible(
        True
    )

    if xlim is not None:
        ax.set_xlim(
            xlim
        )

    if ylim is not None:
        ax.set_ylim(
            ylim
        )

    ax.tick_params(
        axis='x',
        labelrotation=x_rotation,
        labelsize=x_tick_fontsize
    )

    if show_y_axis:
        ax.tick_params(
            axis='y',
            labelrotation=y_rotation,
            labelsize=y_tick_fontsize
        )

    # Preserve spatial proportions
    ax.set_aspect(
        'equal',
        adjustable='box'
    )

    plt.tight_layout()

    # -------------------------------------------------
    # Save
    # -------------------------------------------------
    if save_path is not None:
        fig.savefig(
            save_path,
            dpi=dpi,
            bbox_inches='tight'
        )

    plt.show()

    return centroid_gdf
```

### Just showing the negative values; hide the y-axis 

```python
def plot_site_coef_distances_negative_only(
    gdf,
    value_col='coef_Temp',
    base_gdf=None,
    use_base_df=False,
    ref_easting=None,
    ref_northing=None,
    use_union_centroid=False,
    figsize=(9, 7),
    cbar_label_fontsize=15,
    x_label_fontsize=15,
    x_labelpad=15,
    x_tick_fontsize=12,
    y_label=None,
    y_label_fontsize=15,
    y_labelpad=15,
    y_tick_fontsize=12,
    y_rotation=0,
    show_y_axis=True,

    # Plot styling
    point_size=70,
    point_alpha=0.9,
    annotate=False,
    cbar_label='Voltage Sensitivity to Temperature (ΔV/°C)',
    cbar_labelpad=20,
    x_rotation=45,

    # Background EA basemap
    show_basemap=False,
    basemap_facecolor='none',
    basemap_edgecolor='lightgrey',
    basemap_linewidth=0.4,
    basemap_alpha=1.0,

    # Reference lines
    show_vertical_reference_line=True,
    show_horizontal_reference_line=False,
    reference_linecolor='black',
    reference_linestyle='--',
    reference_linewidth=1,
    reference_alpha=0.6,

    # Panel label using ax.transAxes
    panel_label=None,
    panel_label_x=0.02,
    panel_label_y=0.98,
    panel_label_fontsize=16,
    panel_label_fontweight='bold',
    panel_label_ha='left',
    panel_label_va='top',

    # Distance options
    use_km=True,
    x_ref_offset=0,
    y_ref_offset=0,

    # Axis limits
    xlim=None,
    ylim=None,

    # Site collapsing
    collapse_sites=True,
    collapse_method='mean',

    # Significance filtering
    sig_col=None,
    keep_only_sig=False,

    # Colorbar layout
    colorbar_fraction=0.046,
    colorbar_pad=0.04,
    colorbar_tick_fontsize=10.5,

    # Color scale
    vmin=None,
    vmax=None,

    # Diagnostics
    diagnostics=False,

    # Saving
    save_path=None,
    dpi=300
):
    """
    Plot negative site-level coefficient values relative to a reference
    location.

    Darker red represents more-negative coefficients, while lighter red
    represents coefficients closer to zero.

    Parameters
    ----------
    gdf : GeoDataFrame
        Point GeoDataFrame containing site geometries and coefficients.

    value_col : str
        Column containing coefficient values.

    base_gdf : GeoDataFrame, optional
        Polygon GeoDataFrame used to calculate a common reference point
        and/or display the background basemap.

    use_base_df : bool
        If True, calculate the reference point from the union centroid
        of base_gdf.

    use_union_centroid : bool
        If True and use_base_df is False, calculate the reference point
        from the union centroid of the plotted site geometries.

    show_basemap : bool
        If True, plot base_gdf in coordinates relative to the reference
        location.

    show_vertical_reference_line : bool
        Show the reference easting as a vertical line at x = 0.

    show_horizontal_reference_line : bool
        Show the reference northing as a horizontal line at y = 0.

    x_ref_offset, y_ref_offset : float
        Offsets applied to the calculated reference point. Their units
        are kilometres when use_km=True and metres otherwise.

    Returns
    -------
    GeoDataFrame
        Processed site GeoDataFrame containing easting, northing,
        x_distance, and y_distance.
    """

    import numpy as np
    import geopandas as gpd
    import matplotlib.pyplot as plt

    from shapely import affinity

    # -------------------------------------------------
    # Convert site data to GeoDataFrame
    # -------------------------------------------------
    input_crs = getattr(gdf, 'crs', None)

    gdf = gpd.GeoDataFrame(
        gdf.copy(),
        geometry='geometry',
        crs=input_crs
    )

    if gdf.crs is None:
        gdf = gdf.set_crs(epsg=32630)

    if gdf.crs.is_geographic:
        raise ValueError(
            "gdf must use a projected CRS before calculating "
            "distances. For Greater Accra, use "
            "gdf.to_crs(epsg=32630)."
        )

    required_columns = {
        'site_id',
        value_col,
        'geometry'
    }

    missing_columns = (
        required_columns - set(gdf.columns)
    )

    if missing_columns:
        raise ValueError(
            f"Missing columns in gdf: "
            f"{sorted(missing_columns)}"
        )

    # -------------------------------------------------
    # Optional significance filtering
    # -------------------------------------------------
    if keep_only_sig:
        if sig_col is None:
            raise ValueError(
                "sig_col must be provided when "
                "keep_only_sig=True."
            )

        if sig_col not in gdf.columns:
            raise ValueError(
                f"Significance column '{sig_col}' "
                f"was not found."
            )

        gdf = gdf[
            gdf[sig_col].eq(True)
        ].copy()

    # -------------------------------------------------
    # Clean invalid rows
    # -------------------------------------------------
    gdf[value_col] = (
        gdf[value_col]
        .astype(float)
    )

    gdf = gdf.dropna(
        subset=[
            'site_id',
            value_col,
            'geometry'
        ]
    ).copy()

    gdf = gdf[
        ~gdf.geometry.is_empty
    ].copy()

    gdf = gdf[
        gdf.geometry.is_valid
    ].copy()

    gdf = gdf[
        np.isfinite(gdf[value_col])
    ].copy()

    # -------------------------------------------------
    # Keep only negative coefficients
    # -------------------------------------------------
    gdf = gdf[
        gdf[value_col] < 0
    ].copy()

    if gdf.empty:
        raise ValueError(
            f"No negative values were found in "
            f"'{value_col}'."
        )

    # -------------------------------------------------
    # Collapse multiple rows per site
    # -------------------------------------------------
    if collapse_sites:
        if collapse_method == 'max':
            idx = (
                gdf.groupby('site_id')[value_col]
                .idxmax()
            )

            gdf = (
                gdf.loc[idx]
                .copy()
                .reset_index(drop=True)
            )

        elif collapse_method == 'min':
            idx = (
                gdf.groupby('site_id')[value_col]
                .idxmin()
            )

            gdf = (
                gdf.loc[idx]
                .copy()
                .reset_index(drop=True)
            )

        elif collapse_method == 'mean':
            original_crs = gdf.crs

            agg_dict = {
                value_col: (
                    value_col,
                    'mean'
                ),
                'geometry': (
                    'geometry',
                    'first'
                )
            }

            optional_columns = [
                'cesi_level',
                'site_latitude',
                'site_longitude'
            ]

            for column in optional_columns:
                if column in gdf.columns:
                    agg_dict[column] = (
                        column,
                        'first'
                    )

            if (
                sig_col is not None
                and sig_col in gdf.columns
            ):
                agg_dict[sig_col] = (
                    sig_col,
                    'max'
                )

            gdf = (
                gdf.groupby(
                    'site_id',
                    as_index=False
                )
                .agg(**agg_dict)
            )

            gdf = gpd.GeoDataFrame(
                gdf,
                geometry='geometry',
                crs=original_crs
            )

        else:
            raise ValueError(
                "collapse_method must be 'max', "
                "'min', or 'mean'."
            )

    # -------------------------------------------------
    # Extract site coordinates
    # -------------------------------------------------
    centroid_gdf = gdf.copy()

    centroid_gdf['easting'] = (
        centroid_gdf.geometry.x
    )

    centroid_gdf['northing'] = (
        centroid_gdf.geometry.y
    )

    # -------------------------------------------------
    # Prepare base GeoDataFrame when needed
    # -------------------------------------------------
    prepared_base_gdf = None

    if use_base_df or show_basemap:
        if base_gdf is None:
            raise ValueError(
                "base_gdf must be provided when "
                "use_base_df=True or show_basemap=True."
            )

        base_crs = getattr(
            base_gdf,
            'crs',
            None
        )

        prepared_base_gdf = gpd.GeoDataFrame(
            base_gdf.copy(),
            geometry='geometry',
            crs=base_crs
        )

        if prepared_base_gdf.crs is None:
            prepared_base_gdf = (
                prepared_base_gdf.set_crs(
                    gdf.crs
                )
            )

        if prepared_base_gdf.crs != gdf.crs:
            prepared_base_gdf = (
                prepared_base_gdf.to_crs(
                    gdf.crs
                )
            )

        prepared_base_gdf = (
            prepared_base_gdf
            .dropna(subset=['geometry'])
            .copy()
        )

        prepared_base_gdf = prepared_base_gdf[
            ~prepared_base_gdf.geometry.is_empty
        ].copy()

        prepared_base_gdf = prepared_base_gdf[
            prepared_base_gdf.geometry.is_valid
        ].copy()

        if prepared_base_gdf.empty:
            raise ValueError(
                "No valid geometries were found "
                "in base_gdf."
            )

    # -------------------------------------------------
    # Determine reference point
    # -------------------------------------------------
    if use_base_df:
        try:
            base_geom = (
                prepared_base_gdf
                .geometry
                .union_all()
            )
        except Exception:
            base_geom = (
                prepared_base_gdf
                .geometry
                .unary_union
            )

        ref_point = base_geom.centroid

        ref_easting = ref_point.x
        ref_northing = ref_point.y

    elif use_union_centroid:
        try:
            combined_geom = (
                centroid_gdf
                .geometry
                .union_all()
            )
        except Exception:
            combined_geom = (
                centroid_gdf
                .geometry
                .unary_union
            )

        ref_point = combined_geom.centroid

        ref_easting = ref_point.x
        ref_northing = ref_point.y

    else:
        if ref_easting is None:
            ref_easting = (
                centroid_gdf['easting']
                .mean()
            )

        if ref_northing is None:
            ref_northing = (
                centroid_gdf['northing']
                .mean()
            )

    # -------------------------------------------------
    # Apply reference offsets
    # -------------------------------------------------
    if use_km:
        ref_easting += (
            x_ref_offset * 1000
        )

        ref_northing += (
            y_ref_offset * 1000
        )

        conversion_factor = 1000
        unit_label = 'km'

    else:
        ref_easting += x_ref_offset
        ref_northing += y_ref_offset

        conversion_factor = 1
        unit_label = 'm'

    # -------------------------------------------------
    # Calculate relative distances
    # -------------------------------------------------
    centroid_gdf['x_distance'] = (
        centroid_gdf['easting']
        - ref_easting
    ) / conversion_factor

    centroid_gdf['y_distance'] = (
        centroid_gdf['northing']
        - ref_northing
    ) / conversion_factor

    x_label = (
        f'East–West distance from reference '
        f'({unit_label})'
    )

    # -------------------------------------------------
    # Diagnostics
    # -------------------------------------------------
    if diagnostics:
        print(
            "\n--- Plot diagnostics "
            "(negative-only) ---"
        )

        print(
            "Rows after negative filter:",
            len(centroid_gdf)
        )

        print(
            "Unique sites:",
            centroid_gdf['site_id'].nunique()
        )

        print(
            f"{value_col} range:",
            centroid_gdf[value_col].min(),
            centroid_gdf[value_col].max()
        )

        print(
            "x range:",
            centroid_gdf['x_distance'].min(),
            centroid_gdf['x_distance'].max()
        )

        print(
            "y range:",
            centroid_gdf['y_distance'].min(),
            centroid_gdf['y_distance'].max()
        )

        print(
            "Reference easting:",
            ref_easting
        )

        print(
            "Reference northing:",
            ref_northing
        )

        print(
            "Reference offset:",
            f"x={x_ref_offset} {unit_label},",
            f"y={y_ref_offset} {unit_label}"
        )

        print(
            "Show basemap:",
            show_basemap
        )

        print(
            "Show vertical reference line:",
            show_vertical_reference_line
        )

        print(
            "Show horizontal reference line:",
            show_horizontal_reference_line
        )

        print(
            "xlim:",
            xlim
        )

        print(
            "ylim:",
            ylim
        )

        print(
            "----------------------------------------\n"
        )

    # -------------------------------------------------
    # Define colormap and color limits
    #
    # Reds_r:
    # More negative = darker red
    # Closer to zero = lighter red
    # -------------------------------------------------
    cmap = plt.get_cmap('Reds_r')

    vmin_plot = (
        vmin
        if vmin is not None
        else centroid_gdf[value_col].min()
    )

    vmax_plot = (
        vmax
        if vmax is not None
        else 0.0
    )

    if vmin_plot >= vmax_plot:
        raise ValueError(
            f"Invalid color scale: vmin={vmin_plot} "
            f"must be less than vmax={vmax_plot}."
        )

    # -------------------------------------------------
    # Create figure and axes
    # -------------------------------------------------
    fig, ax = plt.subplots(
        figsize=figsize
    )

    # -------------------------------------------------
    # Optional basemap in relative coordinates
    # -------------------------------------------------
    if show_basemap:
        relative_base = (
            prepared_base_gdf.copy()
        )

        relative_base['geometry'] = (
            relative_base.geometry.apply(
                lambda geom: affinity.scale(
                    affinity.translate(
                        geom,
                        xoff=-ref_easting,
                        yoff=-ref_northing
                    ),
                    xfact=1 / conversion_factor,
                    yfact=1 / conversion_factor,
                    origin=(0, 0)
                )
            )
        )

        relative_base.plot(
            ax=ax,
            facecolor=basemap_facecolor,
            edgecolor=basemap_edgecolor,
            linewidth=basemap_linewidth,
            alpha=basemap_alpha,
            zorder=1
        )

    # -------------------------------------------------
    # Scatter plot
    # -------------------------------------------------
    sc = ax.scatter(
        centroid_gdf['x_distance'],
        centroid_gdf['y_distance'],
        c=centroid_gdf[value_col],
        cmap=cmap,
        vmin=vmin_plot,
        vmax=vmax_plot,
        s=point_size,
        alpha=point_alpha,
        zorder=3
    )

    # -------------------------------------------------
    # Reference lines
    # -------------------------------------------------
    if show_vertical_reference_line:
        ax.axvline(
            0,
            color=reference_linecolor,
            linestyle=reference_linestyle,
            linewidth=reference_linewidth,
            alpha=reference_alpha,
            zorder=2
        )

    if show_horizontal_reference_line:
        ax.axhline(
            0,
            color=reference_linecolor,
            linestyle=reference_linestyle,
            linewidth=reference_linewidth,
            alpha=reference_alpha,
            zorder=2
        )

    # -------------------------------------------------
    # Panel label
    # -------------------------------------------------
    if panel_label is not None:
        ax.text(
            panel_label_x,
            panel_label_y,
            str(panel_label),
            transform=ax.transAxes,
            fontsize=panel_label_fontsize,
            fontweight=panel_label_fontweight,
            horizontalalignment=panel_label_ha,
            verticalalignment=panel_label_va,
            clip_on=False,
            zorder=10
        )

    # -------------------------------------------------
    # Optional site annotations
    # -------------------------------------------------
    if annotate:
        for _, row in centroid_gdf.iterrows():
            ax.text(
                row['x_distance'],
                row['y_distance'],
                str(row['site_id']),
                fontsize=8,
                horizontalalignment='left',
                verticalalignment='bottom',
                zorder=4
            )

    # -------------------------------------------------
    # Colorbar
    # -------------------------------------------------
    cbar = fig.colorbar(
        sc,
        ax=ax,
        fraction=colorbar_fraction,
        pad=colorbar_pad
    )

    cbar.set_label(
        cbar_label,
        fontsize=cbar_label_fontsize,
        fontweight='bold',
        labelpad=cbar_labelpad
    )

    cbar.ax.tick_params(
        labelsize=colorbar_tick_fontsize
    )

    # -------------------------------------------------
    # Axes
    # -------------------------------------------------
    ax.set_xlabel(
        x_label,
        fontsize=x_label_fontsize,
        fontweight='bold',
        labelpad=x_labelpad
    )

    if y_label is None:
        y_label = (
            'North–South distance from reference (km)'
            if use_km
            else 'North–South distance from reference (m)'
        )

    ax.yaxis.set_visible(show_y_axis)

    if show_y_axis:
        ax.set_ylabel(
            y_label,
            fontsize=y_label_fontsize,
            fontweight='bold',
            labelpad=y_labelpad
        )

    # Remove all spines except bottom
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_visible(show_y_axis)
    ax.spines['bottom'].set_visible(True)

    if xlim is not None:
        ax.set_xlim(xlim)

    if ylim is not None:
        ax.set_ylim(ylim)

    ax.tick_params(
        axis='x',
        labelsize=x_tick_fontsize,
        labelrotation=x_rotation
    )

    if show_y_axis:
        ax.tick_params(
            axis='y',
            labelsize=y_tick_fontsize,
            labelrotation=y_rotation
        )

    # Preserve map proportions
    ax.set_aspect(
        'equal',
        adjustable='box'
    )

    plt.tight_layout()

    # -------------------------------------------------
    # Save
    # -------------------------------------------------
    if save_path is not None:
        fig.savefig(
            save_path,
            dpi=dpi,
            bbox_inches='tight'
        )

    plt.show()

    return centroid_gdf
```

```python
def plot_site_coef_bivariate(
    gdf,
    value_col='coef_Temp',
    cesi_col='cesi_level',
    base_gdf=None,
    use_base_df=False,
    ref_easting=None,
    ref_northing=None,
    use_union_centroid=False,
    figsize=(9, 7),
    point_size=70,
    annotate=False,
    x_rotation=45,
    use_km=True,
    x_ref_offset=0,
    y_ref_offset=0,
    xlim=None,
    ylim=None,
    collapse_sites=True,
    collapse_method='mean',
    sig_col=None,
    keep_only_sig=False,
    # --- legend controls ---
    legend_pos=(0.72, 0.03),
    legend_cell_size=0.12,
    legend_x_label='Voltage Sensitivity',
    legend_y_label='CESI',
    legend_coef_labels=('Low', 'High'),
    legend_cesi_labels=('Low', 'High'),
    legend_x_tick_rotation=0,
    legend_x_label_offset=-0.55,
    legend_y_label_offset=-0.55,
    legend_x_tick_offset=-0.1,
    legend_y_tick_offset=-0.1,
    diagnostics=True,
    save_path=None,
    dpi=300
):
    """
    Plots coefficient values using a bivariate colour scheme.

    Two dimensions:
      - coef_Temp magnitude : Low (above median) vs High (at/below median)
      - cesi_level          : Low vs High

    Colour mixing:
      - Coef axis : blue (#2166ac) → red  (#b2182b)
      - CESI axis : light (low)    → dark (high)

    Y-axis is hidden; only a dashed axvline at x=0 is drawn.
    Only values <= 0 are plotted.
    """

    BIVAR_COLORS = {
        ('Low',  'Low'):  '#a6c6e8',
        ('High', 'Low'):  '#f4a582',
        ('Low',  'High'): '#2166ac',
        ('High', 'High'): '#b2182b',
    }

    # ------------------------------------------------------------------
    # Prep
    # ------------------------------------------------------------------
    gdf = gpd.GeoDataFrame(gdf, geometry='geometry').copy()
    if gdf.crs is None:
        gdf = gdf.set_crs(epsg=32630)

    if keep_only_sig and sig_col is not None:
        gdf = gdf[gdf[sig_col] == True].copy()

    gdf = gdf.dropna(subset=[value_col, 'geometry', cesi_col]).copy()
    gdf = gdf[~gdf.geometry.is_empty].copy()
    gdf = gdf[np.isfinite(gdf[value_col])].copy()

    # Keep only zero and negative values
    gdf = gdf[gdf[value_col] <= 0].copy()

    if len(gdf) == 0:
        raise ValueError(f"No values <= 0 found in column '{value_col}'.")

    # ------------------------------------------------------------------
    # Collapse
    # ------------------------------------------------------------------
    if collapse_sites:
        if collapse_method in ('max', 'min'):
            idx = (gdf.groupby('site_id')[value_col].idxmax()
                   if collapse_method == 'max'
                   else gdf.groupby('site_id')[value_col].idxmin())
            gdf = gdf.loc[idx].copy().reset_index(drop=True)

        elif collapse_method == 'mean':
            original_crs = gdf.crs
            agg_dict = {
                value_col:        (value_col, 'mean'),
                cesi_col:         (cesi_col, 'first'),
                'site_latitude':  ('site_latitude', 'first'),
                'site_longitude': ('site_longitude', 'first'),
                'geometry':       ('geometry', 'first'),
            }
            if sig_col is not None and sig_col in gdf.columns:
                agg_dict[sig_col] = (sig_col, 'max')

            gdf = gdf.groupby('site_id').agg(**agg_dict).reset_index()
            gdf = gpd.GeoDataFrame(gdf, geometry='geometry', crs=original_crs)
        else:
            raise ValueError("collapse_method must be 'max', 'min', or 'mean'")

    # ------------------------------------------------------------------
    # Bivariate categories
    # ------------------------------------------------------------------
    median_coef = gdf[value_col].median()
    gdf['coef_cat'] = np.where(gdf[value_col] <= median_coef, 'High', 'Low')
    gdf['cesi_cat'] = gdf[cesi_col].astype(str).str.strip().str.capitalize()
    gdf['bv_color'] = [
        BIVAR_COLORS.get((r['coef_cat'], r['cesi_cat']), '#cccccc')
        for _, r in gdf.iterrows()
    ]

    # ------------------------------------------------------------------
    # Coordinates & reference
    # ------------------------------------------------------------------
    centroid_gdf = gdf.copy()
    centroid_gdf['easting']  = centroid_gdf.geometry.x
    centroid_gdf['northing'] = centroid_gdf.geometry.y

    if use_base_df:
        if base_gdf is None:
            raise ValueError("base_gdf must be provided when use_base_df=True")
        base_gdf = gpd.GeoDataFrame(base_gdf, geometry='geometry').copy()
        if base_gdf.crs is None:
            base_gdf = base_gdf.set_crs(gdf.crs)
        if base_gdf.crs != gdf.crs:
            base_gdf = base_gdf.to_crs(gdf.crs)
        try:
            base_geom = base_gdf.geometry.union_all()
        except Exception:
            base_geom = base_gdf.geometry.unary_union
        ref_easting  = base_geom.centroid.x
        ref_northing = base_geom.centroid.y

    elif use_union_centroid:
        try:
            combined_geom = centroid_gdf.geometry.union_all()
        except Exception:
            combined_geom = centroid_gdf.geometry.unary_union
        ref_easting  = combined_geom.centroid.x
        ref_northing = combined_geom.centroid.y

    else:
        if ref_easting  is None:
            ref_easting  = centroid_gdf['easting'].mean()
        if ref_northing is None:
            ref_northing = centroid_gdf['northing'].mean()

    scale = 1000 if use_km else 1
    ref_easting  += x_ref_offset * scale
    ref_northing += y_ref_offset * scale

    centroid_gdf['x_distance'] = (centroid_gdf['easting']  - ref_easting)
    centroid_gdf['y_distance'] = (centroid_gdf['northing'] - ref_northing)

    if use_km:
        centroid_gdf['x_distance'] /= 1000
        centroid_gdf['y_distance'] /= 1000
        x_label = 'East–West distance from reference (km)'
    else:
        x_label = 'East–West distance from reference (m)'

    # ------------------------------------------------------------------
    # Diagnostics
    # ------------------------------------------------------------------
    if diagnostics:
        print("\n--- Bivariate plot diagnostics ---")
        print(f"Rows plotted : {len(centroid_gdf)}")
        print(f"Median coef  : {median_coef:.4f}")
        print(f"coef_cat counts:\n{centroid_gdf['coef_cat'].value_counts()}")
        print(f"cesi_cat counts:\n{centroid_gdf['cesi_cat'].value_counts()}")

        for cat in ['Low', 'High']:
            sub = centroid_gdf[centroid_gdf['coef_cat'] == cat][value_col]
            if len(sub) > 0:
                print(
                    f"\n'{cat}' sensitivity range: "
                    f"[{sub.min():.4f}, {sub.max():.4f}]  "
                    f"(n={len(sub)})"
                )
            else:
                print(f"\n'{cat}' sensitivity range: no rows")

        print("----------------------------------\n")

    # ------------------------------------------------------------------
    # Plot
    # ------------------------------------------------------------------
    fig, ax = plt.subplots(figsize=figsize)

    ax.scatter(
        centroid_gdf['x_distance'],
        centroid_gdf['y_distance'],
        c=centroid_gdf['bv_color'],
        s=point_size,
        alpha=0.9,
        edgecolors='white',
        linewidths=0.4,
    )

    ax.axvline(0, linestyle='--', linewidth=1, color='black', alpha=0.6)

    ax.yaxis.set_visible(False)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_visible(False)
    ax.spines['bottom'].set_visible(True)

    if annotate:
        for _, row in centroid_gdf.iterrows():
            ax.annotate(
                str(row['site_id']),
                (row['x_distance'], row['y_distance']),
                fontsize=7, ha='left', va='bottom',
            )

    ax.set_xlabel(x_label, fontsize=13, fontweight='bold', labelpad=15)

    if xlim is not None:
        ax.set_xlim(xlim)
    if ylim is not None:
        ax.set_ylim(ylim)

    plt.xticks(rotation=x_rotation)

    # ------------------------------------------------------------------
    # Bivariate inset legend
    # ------------------------------------------------------------------
    legend_bbox = (
        legend_pos[0],
        legend_pos[1],
        legend_cell_size * 2,
        legend_cell_size * 2,
    )

    _add_bivariate_legend(
        ax,
        BIVAR_COLORS,
        bbox=legend_bbox,
        x_label=legend_x_label,
        y_label=legend_y_label,
        coef_labels=legend_coef_labels,
        cesi_labels=legend_cesi_labels,
        x_tick_rotation=legend_x_tick_rotation,
        x_label_offset=legend_x_label_offset,
        y_label_offset=legend_y_label_offset,
        x_tick_offset=legend_x_tick_offset,
        y_tick_offset=legend_y_tick_offset,
    )

    plt.tight_layout()

    if save_path is not None:
        plt.savefig(save_path, dpi=dpi, bbox_inches='tight')

    plt.show()

    return centroid_gdf


# ----------------------------------------------------------------------
# Helper
# ----------------------------------------------------------------------
def _add_bivariate_legend(
    ax, colors, bbox, x_label, y_label,
    coef_labels, cesi_labels,
    x_tick_rotation=0,
    x_label_offset=-0.55,
    y_label_offset=-0.55,
    x_tick_offset=-0.1,
    y_tick_offset=-0.1,
):
    """
    Draws a compact 2×2 bivariate legend as an inset axes.

    Offset params (all in data coords of the inset axes, grid spans 0-2):
      x_label_offset : y position of the x-axis label  (more negative = further below)
      y_label_offset : x position of the y-axis label  (more negative = further left)
      x_tick_offset  : y position of x tick labels     (more negative = further below)
      y_tick_offset  : x position of y tick labels     (more negative = further left)
    """

    axins = ax.inset_axes(bbox)
    axins.set_xlim(0, 2)
    axins.set_ylim(0, 2)
    axins.set_aspect('equal')
    axins.axis('off')

    # row 0 = CESI Low (bottom), row 1 = CESI High (top)
    grid = [
        [('Low',  'Low'),  ('High', 'Low') ],
        [('Low',  'High'), ('High', 'High')],
    ]

    for row_idx, row in enumerate(grid):
        for col_idx, (coef_cat, cesi_cat) in enumerate(row):
            colour = colors.get((coef_cat, cesi_cat), '#cccccc')
            rect = mpatches.FancyBboxPatch(
                (col_idx, row_idx), 1, 1,
                boxstyle='square,pad=0',
                facecolor=colour,
                edgecolor='white',
                linewidth=1.5,
            )
            axins.add_patch(rect)

    fs_label = 10
    fs_tick  = 8.5

    # Axis labels
    axins.text(1.0, x_label_offset, x_label,
               ha='center', va='top', fontsize=fs_label, fontweight='bold',
               transform=axins.transData)
    axins.text(y_label_offset, 1.0, y_label,
               ha='right', va='center', fontsize=fs_label, fontweight='bold',
               rotation=90, transform=axins.transData)

    # X-axis tick labels (coef ranges)
    for col_idx, label in enumerate(coef_labels):
        axins.text(col_idx + 0.5, x_tick_offset, label,
                   ha='right', va='top', fontsize=fs_tick,
                   rotation=x_tick_rotation,
                   rotation_mode='anchor',
                   transform=axins.transData)

    # Y-axis tick labels (cesi ranges)
    for row_idx, label in enumerate(cesi_labels):
        axins.text(y_tick_offset, row_idx + 0.5, label,
                   ha='right', va='center', fontsize=fs_tick,
                   transform=axins.transData)

    # Arrows
    axins.annotate('', xy=(2.15, -0.45), xytext=(0.0, -0.45),
                   arrowprops=dict(arrowstyle='->', color='black', lw=1.0))
    axins.annotate('', xy=(-0.45, 2.15), xytext=(-0.45, 0.0),
                   arrowprops=dict(arrowstyle='->', color='black', lw=1.0))
```

```python
def plot_site_coef_bivariate(
    gdf,
    value_col='coef_Temp',
    cesi_col='cesi_level',
    base_gdf=None,
    use_base_df=False,
    ref_easting=None,
    ref_northing=None,
    use_union_centroid=False,
    figsize=(9, 7),

    # Plot styling
    point_size=70,
    point_alpha=0.9,
    point_edgecolor='white',
    point_linewidth=0.4,
    annotate=False,
    x_rotation=45,
    x_tick_fontsize=12,
    x_label_fontsize=13,
    x_labelpad=15,
    y_label=None,
    y_label_fontsize=13,
    y_labelpad=15,
    y_tick_fontsize=12,
    y_rotation=0,
    show_y_axis=True,

    # Background basemap
    show_basemap=False,
    basemap_facecolor='none',
    basemap_edgecolor='lightgrey',
    basemap_linewidth=0.4,
    basemap_alpha=1.0,

    # Reference lines
    show_vertical_reference_line=True,
    show_horizontal_reference_line=False,
    reference_linecolor='black',
    reference_linestyle='--',
    reference_linewidth=1,
    reference_alpha=0.6,

    # Panel label using ax.transAxes
    panel_label=None,
    panel_label_x=0.02,
    panel_label_y=0.98,
    panel_label_fontsize=16,
    panel_label_fontweight='bold',
    panel_label_ha='left',
    panel_label_va='top',

    # Distance options
    use_km=True,
    x_ref_offset=0,
    y_ref_offset=0,

    # Axis limits
    xlim=None,
    ylim=None,

    # Site collapsing
    collapse_sites=True,
    collapse_method='mean',

    # Significance filtering
    sig_col=None,
    keep_only_sig=False,

    # Quadrant colors
    quadrant_colors=None,

    # Optional Esri web-tile basemap (requires internet)
    esri_basemap=False,
    esri_source=None,
    esri_zoom='auto',
    esri_attribution=None,

    # Bivariate legend controls
    show_legend=True,
    legend_pos=(0.72, 0.03),
    legend_cell_size=0.12,
    legend_x_label='Voltage Sensitivity',
    legend_y_label='CESI',
    legend_coef_labels=('Low', 'High'),
    legend_cesi_labels=('Low', 'High'),
    legend_x_tick_rotation=0,
    legend_x_label_offset=-0.55,
    legend_y_label_offset=-0.55,
    legend_x_tick_offset=-0.1,
    legend_y_tick_offset=-0.1,

    # Diagnostics and saving
    diagnostics=True,
    save_path=None,
    dpi=300
):
    """
    Plot non-positive site-level coefficients using a bivariate scheme.

    The two dimensions are:

    - Coefficient magnitude: Low versus High sensitivity
    - CESI level: Low versus High CESI

    More-negative coefficients are classified as higher voltage
    sensitivity. Locations are expressed relative to a selected
    reference point.

    Reference-point priority
    ------------------------
    1. If use_base_df=True, use the union centroid of base_gdf.
    2. Otherwise, if use_union_centroid=True, use the union centroid
       of the plotted site geometries.
    3. Otherwise, use ref_easting and ref_northing when supplied.
       Any missing coordinate is replaced with the mean site coordinate.

    Returns
    -------
    GeoDataFrame
        Processed site data containing coefficient and CESI categories,
        bivariate colors, and relative coordinates.
    """

    import numpy as np
    import geopandas as gpd
    import matplotlib.pyplot as plt

    from shapely import affinity

    # -------------------------------------------------
    # Default bivariate colors
    # -------------------------------------------------
    default_quadrant_colors = {
        ('Low', 'Low'): '#a6c6e8',
        ('High', 'Low'): '#f4a582',
        ('Low', 'High'): '#2166ac',
        ('High', 'High'): '#b2182b'
    }

    if quadrant_colors is None:
        quadrant_colors = (
            default_quadrant_colors.copy()
        )

    else:
        merged_colors = (
            default_quadrant_colors.copy()
        )

        merged_colors.update(
            quadrant_colors
        )

        quadrant_colors = merged_colors

    # -------------------------------------------------
    # Prepare site GeoDataFrame
    # -------------------------------------------------
    input_crs = getattr(
        gdf,
        'crs',
        None
    )

    gdf = gpd.GeoDataFrame(
        gdf.copy(),
        geometry='geometry',
        crs=input_crs
    )

    if gdf.crs is None:
        gdf = gdf.set_crs(
            epsg=32630
        )

    if gdf.crs.is_geographic:
        raise ValueError(
            'gdf must use a projected CRS before calculating '
            'distances. For Greater Accra, use '
            'gdf.to_crs(epsg=32630).'
        )

    required_columns = {
        'site_id',
        value_col,
        cesi_col,
        'geometry'
    }

    missing_columns = (
        required_columns
        - set(gdf.columns)
    )

    if missing_columns:
        raise ValueError(
            f'Missing columns in gdf: '
            f'{sorted(missing_columns)}'
        )

    # -------------------------------------------------
    # Optional significance filtering
    # -------------------------------------------------
    if keep_only_sig:
        if sig_col is None:
            raise ValueError(
                'sig_col must be provided when '
                'keep_only_sig=True.'
            )

        if sig_col not in gdf.columns:
            raise ValueError(
                f"Significance column '{sig_col}' "
                f'was not found.'
            )

        gdf = gdf[
            gdf[sig_col].eq(True)
        ].copy()

    # -------------------------------------------------
    # Remove invalid observations
    # -------------------------------------------------
    gdf[value_col] = (
        gdf[value_col]
        .astype(float)
    )

    gdf = gdf.dropna(
        subset=[
            'site_id',
            value_col,
            cesi_col,
            'geometry'
        ]
    ).copy()

    gdf = gdf[
        ~gdf.geometry.is_empty
    ].copy()

    gdf = gdf[
        gdf.geometry.is_valid
    ].copy()

    gdf = gdf[
        np.isfinite(gdf[value_col])
    ].copy()

    # Keep zero and negative coefficients
    gdf = gdf[
        gdf[value_col] <= 0
    ].copy()

    if gdf.empty:
        raise ValueError(
            f"No values <= 0 were found in "
            f"'{value_col}'."
        )

    # -------------------------------------------------
    # Collapse multiple rows per site
    # -------------------------------------------------
    if collapse_sites:
        if collapse_method in (
            'max',
            'min'
        ):
            grouped = (
                gdf.groupby('site_id')[value_col]
            )

            if collapse_method == 'max':
                idx = grouped.idxmax()

            else:
                idx = grouped.idxmin()

            gdf = (
                gdf.loc[idx]
                .copy()
                .reset_index(drop=True)
            )

        elif collapse_method == 'mean':
            original_crs = gdf.crs

            agg_dict = {
                value_col: (
                    value_col,
                    'mean'
                ),
                cesi_col: (
                    cesi_col,
                    'first'
                ),
                'geometry': (
                    'geometry',
                    'first'
                )
            }

            optional_columns = [
                'site_latitude',
                'site_longitude'
            ]

            for column in optional_columns:
                if column in gdf.columns:
                    agg_dict[column] = (
                        column,
                        'first'
                    )

            if (
                sig_col is not None
                and sig_col in gdf.columns
            ):
                agg_dict[sig_col] = (
                    sig_col,
                    'max'
                )

            gdf = (
                gdf.groupby(
                    'site_id',
                    as_index=False
                )
                .agg(**agg_dict)
            )

            gdf = gpd.GeoDataFrame(
                gdf,
                geometry='geometry',
                crs=original_crs
            )

        else:
            raise ValueError(
                "collapse_method must be 'max', "
                "'min', or 'mean'."
            )

    # -------------------------------------------------
    # Create bivariate categories
    # -------------------------------------------------
    median_coef = (
        gdf[value_col].median()
    )

    # More-negative coefficients represent higher
    # temperature sensitivity.
    gdf['coef_cat'] = np.where(
        gdf[value_col] <= median_coef,
        'High',
        'Low'
    )

    gdf['cesi_cat'] = (
        gdf[cesi_col]
        .astype(str)
        .str.strip()
        .str.title()
    )

    gdf['bv_color'] = [
        quadrant_colors.get(
            (
                row['coef_cat'],
                row['cesi_cat']
            ),
            '#cccccc'
        )
        for _, row in gdf.iterrows()
    ]

    # -------------------------------------------------
    # Site coordinates
    # -------------------------------------------------
    centroid_gdf = gdf.copy()

    centroid_gdf['easting'] = (
        centroid_gdf.geometry.x
    )

    centroid_gdf['northing'] = (
        centroid_gdf.geometry.y
    )

    # -------------------------------------------------
    # Prepare basemap when needed
    # -------------------------------------------------
    prepared_base_gdf = None

    if use_base_df or show_basemap:
        if base_gdf is None:
            raise ValueError(
                'base_gdf must be provided when '
                'use_base_df=True or show_basemap=True.'
            )

        base_crs = getattr(
            base_gdf,
            'crs',
            None
        )

        prepared_base_gdf = gpd.GeoDataFrame(
            base_gdf.copy(),
            geometry='geometry',
            crs=base_crs
        )

        if prepared_base_gdf.crs is None:
            prepared_base_gdf = (
                prepared_base_gdf.set_crs(
                    gdf.crs
                )
            )

        if prepared_base_gdf.crs != gdf.crs:
            prepared_base_gdf = (
                prepared_base_gdf.to_crs(
                    gdf.crs
                )
            )

        prepared_base_gdf = (
            prepared_base_gdf
            .dropna(subset=['geometry'])
            .copy()
        )

        prepared_base_gdf = prepared_base_gdf[
            ~prepared_base_gdf.geometry.is_empty
        ].copy()

        prepared_base_gdf = prepared_base_gdf[
            prepared_base_gdf.geometry.is_valid
        ].copy()

        if prepared_base_gdf.empty:
            raise ValueError(
                'No valid geometries were found '
                'in base_gdf.'
            )

    # -------------------------------------------------
    # Determine reference point
    # -------------------------------------------------
    if use_base_df:
        try:
            base_geom = (
                prepared_base_gdf
                .geometry
                .union_all()
            )

        except Exception:
            base_geom = (
                prepared_base_gdf
                .geometry
                .unary_union
            )

        reference_point = (
            base_geom.centroid
        )

        ref_easting = (
            reference_point.x
        )

        ref_northing = (
            reference_point.y
        )

        reference_source = (
            'base_gdf union centroid'
        )

    elif use_union_centroid:
        try:
            combined_geom = (
                centroid_gdf
                .geometry
                .union_all()
            )

        except Exception:
            combined_geom = (
                centroid_gdf
                .geometry
                .unary_union
            )

        reference_point = (
            combined_geom.centroid
        )

        ref_easting = (
            reference_point.x
        )

        ref_northing = (
            reference_point.y
        )

        reference_source = (
            'site union centroid'
        )

    else:
        if ref_easting is None:
            ref_easting = (
                centroid_gdf['easting']
                .mean()
            )

        if ref_northing is None:
            ref_northing = (
                centroid_gdf['northing']
                .mean()
            )

        reference_source = (
            'supplied coordinates or site means'
        )

    # -------------------------------------------------
    # Apply reference offsets
    # -------------------------------------------------
    if use_km:
        conversion_factor = 1000
        unit_label = 'km'

    else:
        conversion_factor = 1
        unit_label = 'm'

    ref_easting += (
        x_ref_offset
        * conversion_factor
    )

    ref_northing += (
        y_ref_offset
        * conversion_factor
    )

    # -------------------------------------------------
    # Calculate relative coordinates
    # -------------------------------------------------
    centroid_gdf['x_distance'] = (
        centroid_gdf['easting']
        - ref_easting
    ) / conversion_factor

    centroid_gdf['y_distance'] = (
        centroid_gdf['northing']
        - ref_northing
    ) / conversion_factor

    x_label = (
        f'East–West distance from reference '
        f'({unit_label})'
    )

    # -------------------------------------------------
    # Diagnostics
    # -------------------------------------------------
    if diagnostics:
        print(
            '\n--- Bivariate plot diagnostics ---'
        )

        print(
            f'Rows plotted: {len(centroid_gdf)}'
        )

        print(
            'Unique sites:',
            centroid_gdf['site_id'].nunique()
        )

        print(
            f'Median coefficient: '
            f'{median_coef:.4f}'
        )

        print(
            f'Reference source: '
            f'{reference_source}'
        )

        print(
            f'Reference easting: '
            f'{ref_easting}'
        )

        print(
            f'Reference northing: '
            f'{ref_northing}'
        )

        print(
            f'Reference offset: '
            f'x={x_ref_offset} {unit_label}, '
            f'y={y_ref_offset} {unit_label}'
        )

        print(
            '\nCoefficient categories:\n',
            centroid_gdf[
                'coef_cat'
            ].value_counts()
        )

        print(
            '\nCESI categories:\n',
            centroid_gdf[
                'cesi_cat'
            ].value_counts()
        )

        for category in [
            'Low',
            'High'
        ]:
            subset = centroid_gdf.loc[
                centroid_gdf[
                    'coef_cat'
                ].eq(category),
                value_col
            ]

            if not subset.empty:
                print(
                    f"\n'{category}' sensitivity range: "
                    f'[{subset.min():.4f}, '
                    f'{subset.max():.4f}] '
                    f'(n={len(subset)})'
                )

        print(
            '\nx range:',
            centroid_gdf[
                'x_distance'
            ].min(),
            centroid_gdf[
                'x_distance'
            ].max()
        )

        print(
            'y range:',
            centroid_gdf[
                'y_distance'
            ].min(),
            centroid_gdf[
                'y_distance'
            ].max()
        )

        print(
            'Show basemap:',
            show_basemap
        )

        print(
            'Show vertical reference line:',
            show_vertical_reference_line
        )

        print(
            'Show horizontal reference line:',
            show_horizontal_reference_line
        )

        print(
            '----------------------------------\n'
        )

    # -------------------------------------------------
    # Create figure
    # -------------------------------------------------
    fig, ax = plt.subplots(
        figsize=figsize
    )

    # -------------------------------------------------
    # Optional polygon and Esri basemaps
    # -------------------------------------------------
    if show_basemap:
        relative_base = prepared_base_gdf.copy()
        if not esri_basemap:
            relative_base['geometry'] = (
                relative_base.geometry.apply(
                    lambda geometry: affinity.scale(
                        affinity.translate(
                            geometry,
                            xoff=-ref_easting,
                            yoff=-ref_northing
                        ),
                        xfact=1 / conversion_factor,
                        yfact=1 / conversion_factor,
                        origin=(0, 0)
                    )
                )
            )

        relative_base.plot(
            ax=ax,
            facecolor=basemap_facecolor,
            edgecolor=basemap_edgecolor,
            linewidth=basemap_linewidth,
            alpha=basemap_alpha,
            zorder=1
        )

    plot_x = (centroid_gdf['easting'] if esri_basemap
              else centroid_gdf['x_distance'])
    plot_y = (centroid_gdf['northing'] if esri_basemap
              else centroid_gdf['y_distance'])

    # -------------------------------------------------
    # Scatter plot
    # -------------------------------------------------
    ax.scatter(
        plot_x,
        plot_y,
        c=centroid_gdf['bv_color'],
        s=point_size,
        alpha=point_alpha,
        edgecolors=point_edgecolor,
        linewidths=point_linewidth,
        zorder=3
    )

    if esri_basemap:
        if xlim is not None:
            ax.set_xlim(tuple(
                ref_easting + value * conversion_factor for value in xlim
            ))
        if ylim is not None:
            ax.set_ylim(tuple(
                ref_northing + value * conversion_factor for value in ylim
            ))

        try:
            import contextily as ctx
        except ImportError as exc:
            raise ImportError(
                'Esri tiles require contextily. Install it with '
                '`conda install -c conda-forge contextily` or '
                '`pip install contextily`.'
            ) from exc

        source = (ctx.providers.Esri.WorldGrayCanvas
                  if esri_source is None else esri_source)
        ctx.add_basemap(
            ax, source=source, crs=centroid_gdf.crs,
            zoom=esri_zoom, attribution=esri_attribution,
            reset_extent=True
        )

    # -------------------------------------------------
    # Reference lines
    # -------------------------------------------------
    if show_vertical_reference_line:
        ax.axvline(
            ref_easting if esri_basemap else 0,
            color=reference_linecolor,
            linestyle=reference_linestyle,
            linewidth=reference_linewidth,
            alpha=reference_alpha,
            zorder=2
        )

    if show_horizontal_reference_line:
        ax.axhline(
            ref_northing if esri_basemap else 0,
            color=reference_linecolor,
            linestyle=reference_linestyle,
            linewidth=reference_linewidth,
            alpha=reference_alpha,
            zorder=2
        )

    # -------------------------------------------------
    # Panel label
    # -------------------------------------------------
    if panel_label is not None:
        ax.text(
            panel_label_x,
            panel_label_y,
            str(panel_label),
            transform=ax.transAxes,
            fontsize=panel_label_fontsize,
            fontweight=panel_label_fontweight,
            horizontalalignment=panel_label_ha,
            verticalalignment=panel_label_va,
            clip_on=False,
            zorder=10
        )

    # -------------------------------------------------
    # Optional site annotations
    # -------------------------------------------------
    if annotate:
        for _, row in centroid_gdf.iterrows():
            ax.annotate(
                str(row['site_id']),
                (
                    row['easting'] if esri_basemap else row['x_distance'],
                    row['northing'] if esri_basemap else row['y_distance']
                ),
                fontsize=7,
                horizontalalignment='left',
                verticalalignment='bottom',
                zorder=4
            )

    # -------------------------------------------------
    # Axes
    # -------------------------------------------------
    ax.set_xlabel(
        x_label,
        fontsize=x_label_fontsize,
        fontweight='bold',
        labelpad=x_labelpad
    )

    if y_label is None:
        y_label = (
            'North–south distance from reference (km)'
            if use_km
            else 'North–south distance from reference (m)'
        )

    ax.yaxis.set_visible(show_y_axis)

    if show_y_axis:
        ax.set_ylabel(
            y_label,
            fontsize=y_label_fontsize,
            fontweight='bold',
            labelpad=y_labelpad
        )

    ax.spines['top'].set_visible(
        False
    )

    ax.spines['right'].set_visible(
        False
    )

    ax.spines['left'].set_visible(
        show_y_axis
    )

    ax.spines['bottom'].set_visible(
        True
    )

    if xlim is not None:
        ax.set_xlim(
            tuple(ref_easting + value * conversion_factor for value in xlim)
            if esri_basemap else xlim
        )

    if ylim is not None:
        ax.set_ylim(
            tuple(ref_northing + value * conversion_factor for value in ylim)
            if esri_basemap else ylim
        )

    if esri_basemap:
        from matplotlib.ticker import FuncFormatter
        ax.xaxis.set_major_formatter(FuncFormatter(
            lambda value, position: f'{(value - ref_easting) / conversion_factor:g}'
        ))
        ax.yaxis.set_major_formatter(FuncFormatter(
            lambda value, position: f'{(value - ref_northing) / conversion_factor:g}'
        ))

    ax.tick_params(
        axis='x',
        labelsize=x_tick_fontsize,
        labelrotation=x_rotation
    )

    if show_y_axis:
        ax.tick_params(
            axis='y',
            labelsize=y_tick_fontsize,
            labelrotation=y_rotation
        )

    # Preserve spatial proportions
    ax.set_aspect(
        'equal',
        adjustable='box'
    )

    # -------------------------------------------------
    # Bivariate inset legend
    # -------------------------------------------------
    if show_legend:
        legend_bbox = (
            legend_pos[0],
            legend_pos[1],
            legend_cell_size * 2,
            legend_cell_size * 2
        )

        _add_bivariate_legend(
            ax=ax,
            colors=quadrant_colors,
            bbox=legend_bbox,
            x_label=legend_x_label,
            y_label=legend_y_label,
            coef_labels=legend_coef_labels,
            cesi_labels=legend_cesi_labels,
            x_tick_rotation=legend_x_tick_rotation,
            x_label_offset=legend_x_label_offset,
            y_label_offset=legend_y_label_offset,
            x_tick_offset=legend_x_tick_offset,
            y_tick_offset=legend_y_tick_offset
        )

    plt.tight_layout()

    # -------------------------------------------------
    # Save
    # -------------------------------------------------
    if save_path is not None:
        fig.savefig(
            save_path,
            dpi=dpi,
            bbox_inches='tight'
        )

    plt.show()

    return centroid_gdf


def _add_bivariate_legend(
    ax,
    colors,
    bbox,
    x_label,
    y_label,
    coef_labels,
    cesi_labels,
    x_tick_rotation=0,
    x_label_offset=-0.15,
    y_label_offset=-0.15,
    x_tick_offset=-0.1,
    y_tick_offset=-0.1
):
    """Draw a compact 2 × 2 bivariate legend as an inset axes."""

    import matplotlib.patches as mpatches

    axins = ax.inset_axes(
        bbox
    )

    axins.set_xlim(
        0,
        2
    )

    axins.set_ylim(
        0,
        2
    )

    axins.set_aspect(
        'equal'
    )

    axins.axis(
        'off'
    )

    # Bottom row: low CESI
    # Top row: high CESI
    grid = [
        [
            ('Low', 'Low'),
            ('High', 'Low')
        ],
        [
            ('Low', 'High'),
            ('High', 'High')
        ]
    ]

    for row_index, row in enumerate(grid):
        for column_index, categories in enumerate(row):
            coefficient_category, cesi_category = (
                categories
            )

            color = colors.get(
                (
                    coefficient_category,
                    cesi_category
                ),
                '#cccccc'
            )

            rectangle = (
                mpatches.FancyBboxPatch(
                    (
                        column_index,
                        row_index
                    ),
                    1,
                    1,
                    boxstyle='square,pad=0',
                    facecolor=color,
                    edgecolor='white',
                    linewidth=1.5
                )
            )

            axins.add_patch(
                rectangle
            )

    label_fontsize = 11
    tick_fontsize = 9

    # Axis labels
    axins.text(
        1,
        x_label_offset,
        x_label,
        horizontalalignment='center',
        verticalalignment='top',
        fontsize=label_fontsize,
        fontweight='bold',
        transform=axins.transData
    )

    axins.text(
        y_label_offset,
        1,
        y_label,
        horizontalalignment='right',
        verticalalignment='center',
        fontsize=label_fontsize,
        fontweight='bold',
        rotation=90,
        transform=axins.transData
    )

    # Coefficient labels
    for column_index, label in enumerate(
        coef_labels
    ):
        axins.text(
            column_index + 0.5,
            x_tick_offset,
            label,
            horizontalalignment='right',
            verticalalignment='top',
            fontsize=tick_fontsize,
            rotation=x_tick_rotation,
            rotation_mode='anchor',
            transform=axins.transData
        )

    # CESI labels
    for row_index, label in enumerate(
        cesi_labels
    ):
        axins.text(
            y_tick_offset,
            row_index + 0.5,
            label,
            horizontalalignment='right',
            verticalalignment='center',
            fontsize=tick_fontsize,
            transform=axins.transData
        )

    # Direction arrows
    axins.annotate(
        '',
        xy=(2.15, -0.45),
        xytext=(0, -0.45),
        arrowprops={
            'arrowstyle': '->',
            'color': 'black',
            'lw': 1
        }
    )

    axins.annotate(
        '',
        xy=(-0.45, 2.15),
        xytext=(-0.45, 0),
        arrowprops={
            'arrowstyle': '->',
            'color': 'black',
            'lw': 1
        }
    )
```
